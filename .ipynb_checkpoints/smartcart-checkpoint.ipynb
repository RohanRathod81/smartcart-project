```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("smartcart_customers.csv")
```

```python
df.shape
```

```python
# Looking at data

df.isnull().sum()

df.info()
```

## Pre-processing data

1. Handle Missing Values

```python
df["Income"] = df["Income"].fillna(df["Income"].median())
```

```python
df.head()
df.isnull().sum()
```

Feature Engineering

```python
df.columns
```

Age

```python
# Age = sys().year() - df["Year_Birth"]

df["Age"] = 2026-df["Year_Birth"]
```

Customer_Tenure_Days

```python
#Dt_Customer using to cal Customer_Tenure_Days
# Customer joining date


# pandas read csv Dt_custromer as string, so convert to date
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], dayfirst=True)

reference_date = df["Dt_Customer"].max()

df["Customer_Tenure_Days"] = (reference_date - df["Dt_Customer"]).dt.days
```

```python
df.columns
```

Total Spending & Total Children

```python
# Total Spending by customer = wines+ fruits+meats+fish+sweets+gold

df["Total_Spending"] =df["MntWines"] + df["MntFruits"] + df["MntMeatProducts"] + df["MntFishProducts"] + df["MntSweetProducts"] + df["MntGoldProds"] 
```

```python
df.columns
```

```python
# Total Children

df["Total_Children"] =df["Kidhome"] + df["Teenhome"]
```

Education into 3 Groups Replace - UnderGraduation / Graduation /  Post-Graduation

```python
# Education - Dividing into 3 category - UnderGraduation / Graduation /  Post-Graduation

df["Education"].value_counts()

df["Education"] = df["Education"].replace({
    "Basic": "Undergraduate", "2n Cycle": "Undergraduate",
    "Graduation": "Graduate", 
    "Master" : "Postgraduate", "PhD" : "Postgraduate"
})
```

```python
df.head()
```

Marital_Status - Replace with 2 groups - Partner / Alone

```python
# Matrial Status - Dividing into 2 category - Partner / Alone

df["Marital_Status"].value_counts()

df["Living_With"] = df["Marital_Status"].replace({
    "Married": "Partner", "Together": "Partner",
    "Single": "Alone", "Divorced" : "Alone",
    "Widow" : "Alone", "Absurd" : "Alone", "YOLO": "Alone"
})
```

```python
df["Living_With"].value_counts()
```

Drop Columns

```python
# Before drop = df.shape = (2240,27)

df.head()
df.shape
```

```python
cols = ["ID", "Year_Birth", "Marital_Status", "Kidhome", "Teenhome", "Dt_Customer"]
spending_cols = ["MntWines", "MntFruits", "MntMeatProducts", "MntFishProducts","MntSweetProducts", "MntGoldProds"]

cols_to_drop = cols + spending_cols

df_cleaned = df.drop(columns = cols_to_drop)
```

```python
# After dropping cols = df_cleaned.shape = (2240,15)

df_cleaned.head()
df_cleaned.shape
```

Outliers

```python
df_cleaned.columns
```

```python
cols = ["Income", "Recency", "Response", "Age", "Total_Spending", "Total_Children"]

# Relative plots of some features - pair-plots
sns.pairplot(df_cleaned[cols])
```

```python
## Removing outlier 

print("data size with outliers:", len(df_cleaned))

df_cleaned = df_cleaned[ (df_cleaned["Age"] < 90) ]
df_cleaned = df_cleaned[(df_cleaned["Income"] < 600_0000)]

print("data size without outlier:", len(df_cleaned))
```

Encoding

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder()

cat_cols = ["Education", "Living_With"]

enc_cols = ohe.fit_transform(df_cleaned[cat_cols])
```

```python
enc_df = pd.DataFrame(enc_cols.toarray(), columns=ohe.get_feature_names_out(cat_cols), index=df_cleaned.index)
```

```python
enc_df
```

```python
df_encoded= pd.concat([df_cleaned.drop(columns= cat_cols), enc_df], axis=1)
```

```python
df_encoded.shape
```

```python
df_encoded.head()
```

Scaling

```python
from sklearn.preprocessing import StandardScaler

X = df_encoded
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

Visualize 

```python
X_scaled.shape
```

```python
# Visualzing the data in 2D & 3D using PCA

from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=3)

X_pca = pca.fit_transform(X_scaled)
```

```python
# PLOT in 2D

plt.scatter(X_pca[:,0], X_pca[:,1])
```

```python
# CAPTURING less Variance in 2D = 0.22 + 0.11 = 0.33

# CAPTURING  Variance in 3D = 0.221 + 0.113 = 0.104 = 0.43

pca.explained_variance_ratio_
```

```python
# PLOT in 3D 

fig = plt.figure(figsize=(8,6))

ax =fig.add_subplot(111, projection = "3d")
ax.scatter(X_pca[:,0],X_pca[:,1], X_pca[:,2] )

ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
ax.set_title("3d-projection")
```

Analyze K value

1. Elbow Method using wcss

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator

wcss = []
for k in range(1, 11):
    kmeans =KMeans(n_clusters=k, random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
```

```python
knee =KneeLocator(range(1,11), wcss, curve="convex", direction="decreasing")
optimal_k =knee.elbow
```

```python
print("best k :", optimal_k)
```

```python
#PLot

plt.plot(range(1,11), wcss, marker="o")
plt.xlabel("K")
plt.ylabel("WCSS")
```

2. Silhouette Score

```python
from sklearn.metrics import silhouette_score

scores = []

for k in range(2,11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels =kmeans.fit_predict(X_pca)
    score= silhouette_score(X_pca, labels)
    scores.append(score)
```

```python
# plot
plt.plot(range(2,11), scores, marker='o')
plt.xlabel("K")
plt.ylabel("Silhouette_score")
```

```python
# combined plot

k_range = range(2,11)

fig, ax1=plt.subplots(figsize=(8,6))

ax1.plot(k_range, wcss[:len(k_range)], marker="o", color ="blue")
ax1.set_xlabel("K")
ax1.set_ylabel("WCSS")

ax2 = ax1.twinx()
ax2.plot(k_range, scores[:len(k_range)], marker="x", color="red", linestyle="--")
ax2.set_ylabel("SS")
```

Clustering

```python
# K Means

kmeans = KMeans(n_clusters=4, random_state=42)
labels_kmeans = kmeans.fit_predict(X_pca)
```

```python
# plot


fig = plt.figure(figsize=(8,6))

ax =fig.add_subplot(111, projection = "3d")
ax.scatter(X_pca[:,0],X_pca[:,1], X_pca[:,2], c=labels_kmeans )
```

Agglomerative Clustering - Hirarchical Clustering

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
agg_clf = AgglomerativeClustering(n_clusters=4, linkage="ward")
labels_agg = agg_clf.fit_predict(X_pca)
```

```python
# plot


fig = plt.figure(figsize=(8,6))

ax =fig.add_subplot(111, projection = "3d")
ax.scatter(X_pca[:,0],X_pca[:,1], X_pca[:,2], c=labels_agg )
```

Characterization of Clusters

```python
# Added labels_agg col in our cleaned df

df_cleaned.drop("labels", axis=1)

df_cleaned["cluster"] = labels_agg
```

```python
df_cleaned.head()
```

```python
pal = ["red", "blue", "yellow", "green"]

sns.countplot(x=df_cleaned["cluster"], palette=pal, hue=df_cleaned["cluster"])
```
